```python
def run_scored(batsman_name):
    vk=data[data['batsman']==batsman_name]
    return vk.groupby('bowling_team')['batsman_runs'].sum().sort_values(ascending=False).head(10)
    #this will give how much a batsman score against a team (batsman top 10 best scores)
```

```python
run_scored('S Dhawan')
```

```python
data2=pd.read_csv('matches.csv')
```
